```python
import pandas as pd
import matplotlib.pyplot as plt

data={
    "Monster":["Dragon", "Ghost", "Robot", "Zombie", "Alien", "UFO"],
    "Power":[95, 60, 25, 15, 10, 45],
    "Speed":[70, 95, 60, 30, 80,45],
    "IQ":[80, 75, 100, 20, 160, 25]
}


df= pd.DataFrame(data)
print(df)


```

```python
print(df["Power"].max())
print(df["Power"].mean())
print(df.sort_values("Power", ascending=False))
```

```python
colors=["red","blue","gray","green","purple","pink"]
plt.bar(df["Monster"], df["Power"],color=colors)
plt.title("Monster Power")
plt.xlabel("Monster")
plt.ylabel("Power")
plt.show()
```

```python
df.plot(
    x="Monster",
    y=["Power", "Speed", "IQ"],
    kind="bar",
    figsize=(10,5)
)
plt.title("Monster Info")
plt.ylabel("Score")
plt.ylim(0,200)
plt.grid(axis="y")
```

```python
plt.plot(df["Monster"], df["Speed"], marker="o", color="red")
plt.show()
```

```python
df["Total"]=(
    df["Power"]+df["Speed"]+df["IQ"]
)
print(df)
```

```python
df["Total"].idxmax()
```

```python
print(df.loc[df["Total"].idxmax(),"Monster"])
```
