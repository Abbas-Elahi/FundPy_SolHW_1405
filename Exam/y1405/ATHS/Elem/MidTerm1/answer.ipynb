```python
import pandas as pd

df = pd.read_csv("world_foods.csv")

print(df.head())
print(df.tail(3))
```

```python
print(df.shape)

print(df.columns)

print(df.columns.to_list())
```

```python
df.info()
```

```python
print(df.describe())
```

```python
total_orders = df["orders"].sum()

print(total_orders)
```

```python
average_price = df["price_usd"].mean()

print(average_price)
```

```python
max_price = df["price_usd"].max()

min_price = df["price_usd"].min()

print(max_price)
print(min_price)
```

```python
average_rating = df["rating"].mean()

print(average_rating)
```

```python
print(df["price_usd"].describe())
```

```python
italian_foods = df[
    (df["country"] == "Italy") &
    (df["price_usd"] < 15)
]

print(
    italian_foods[
        ["dish", "restaurant", "price_usd"]
    ]
)
```

```python
print(italian_foods.shape)
```

```python
cheap_popular = df[
    (df["price_usd"] < 10) &
    (df["rating"] > 4.5)
]

print(cheap_popular)
```

```python
print(cheap_popular.shape)
```

```python
japanese_or_dessert = df[
    (df["country"] == "Japan") |
    (df["category"] == "Dessert")
]

print(japanese_or_dessert)
```

```python
print(japanese_or_dessert.shape)
```

```python
best_sellers = df.sort_values(
    "orders",
    ascending=False
)

print(
    best_sellers[
        ["dish", "country", "restaurant", "orders"]
    ].head(5)
)
```

```python
expensive_foods = df.sort_values(
    "price_usd",
    ascending=False
)

print(
    expensive_foods[
        ["dish", "country", "price_usd"]
    ].head(3)
)
```

```python
country_orders = df.groupby(
    "country"
)["orders"].sum()

country_orders = country_orders.sort_values(
    ascending=False
)

print(country_orders)
```

```python
country_report = df.groupby("country").agg({
    "dish": "count",
    "orders": "sum",
    "price_usd": "mean",
    "rating": "mean"
})

print(country_report)
```

```python
print(
    country_report.sort_values(
        "orders",
        ascending=False
    ).head(1)
)
```

```python
print(
    country_report.sort_values(
        "price_usd",
        ascending=False
    ).head(1)
)
```

```python
df["revenue_usd"] = (
    df["price_usd"] * df["orders"]
)

print(df.head())
```

```python
print(df.shape)
```

```python
top_revenue = df.sort_values(
    "revenue_usd",
    ascending=False
)

print(
    top_revenue[
        [
            "dish",
            "country",
            "price_usd",
            "orders",
            "revenue_usd"
        ]
    ].head(5)
)
```

```python
country_revenue = df.groupby(
    "country"
)["revenue_usd"].sum()

country_revenue = country_revenue.sort_values(
    ascending=False
)

print(country_revenue)
```

```python
new_food = pd.DataFrame([
    {
        "id": 151,
        "dish": "Chicken Tahchin",
        "country": "Iran",
        "restaurant": "Persian Flame",
        "category": "Main",
        "price_usd": 14,
        "orders": 180,
        "rating": 4.8
    }
])
```

```python
df = pd.concat(
    [df, new_food],
    ignore_index=True
)

print(df.tail(3))
```

```python
print(df.shape)
```

```python
df.loc[
    df["id"] == 1,
    "price_usd"
] = 16
```

```python
print(
    df[df["id"] == 1]
)
```

```python
df = df[df["id"] != 150]
```

```python
df = df.reset_index(drop=True)

print(df.tail(3))
```

```python
print(df.shape)
```

```python
df["revenue_usd"] = (
    df["price_usd"] * df["orders"]
)
```

```python
print(
    df[
        (df["id"] == 1) |
        (df["id"] == 151)
    ][
        ["id", "dish", "revenue_usd"]
    ]
)
```

```python
df.to_csv(
    "world_foods_updated.csv",
    index=False
)
```

```python
final_report = df.groupby("country").agg({
    "orders": "sum",
    "revenue_usd": "sum",
    "price_usd": "mean"
})

print(final_report)
```

```python
final_report.reset_index().to_csv(
    "country_report.csv",
    index=False
)
```

```python
special_menu = df[
    (df["price_usd"] < 15) &
    (df["rating"] > 4.6) &
    (df["orders"] > 300)
]

print(special_menu)
```

```python
special_menu = special_menu.sort_values(
    "orders",
    ascending=False
)

print(special_menu)
```

```python
top_special_menu = special_menu.head(5)

print(
    top_special_menu[
        [
            "dish",
            "country",
            "price_usd",
            "orders",
            "rating"
        ]
    ]
)
```
